```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
```

```python
X = wine['data']
```

```python
X.shape
```

**Step 1**

```python
mu = X.mean(axis=0)
X_ = (X - mu)
X_.shape
```

**Step 2**

```python
S = np.dot(X_.T, X_)
S.shape
```

**Step 3**

```python
lambdas, vectors = np.linalg.eig(S)
```

```python
lambdas.shape
```

```python
lambdas[:2]
```

```python
vectors.shape
```

**Step 4**

```python
V = vectors[:, :2]
V.shape
```

**Projection**

```python
X_.shape
```

```python
V.shape
```

```python
X_new = np.dot(X_, V)
X_new.shape
```

```python
X_new[:5]
```

**Sklearn**

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
skX_ = pca.fit_transform(X)
skX_.shape
```

```python
skX_[:5]
```

```python
np.allclose(X_new, -skX_)
```
